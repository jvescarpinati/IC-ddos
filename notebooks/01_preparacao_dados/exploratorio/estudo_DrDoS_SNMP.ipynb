```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('../../../datasets/CICDDOS2019/01-12/DrDoS_SNMP.csv')
pd.set_option('display.max_columns', None)

df.columns = df.columns.str.strip()

df['Label'].value_counts()
```

```python
amostra1 = df[df['Label'] == 'BENIGN'].sample()
amostra2 = df[df['Label'] == 'DrDoS_SNMP'].sample()

amostra = pd.concat([amostra1, amostra2])
display(amostra)
```

```python
cont1 = df['Protocol'].value_counts()
cont2 = df[df['Label'] == 'DrDoS_SNMP']['Protocol'].value_counts()

print(cont1)
print('\n')
print(cont2)
```

```python
amostra3 = df[(df['Protocol'] == 6) & (df['Label'] == 'DrDoS_SNMP')].sample()
display(amostra3)
```

```python
analise = ['Flow Duration', 'Down/Up Ratio', 'Fwd Packet Length Mean', 'Inbound', 'Flow Bytes/s', 'Flow Packets/s', 'Total Backward Packets', 'Total Fwd Packets']

pd.set_option('display.max_rows', None)
display(df.groupby('Label')[analise].describe().T)
```

```python
df_numerico = df.select_dtypes(include=[np.number])

colunas_negativas = df_numerico.columns[(df_numerico < 0).any()]

for col in colunas_negativas:
    qtd = (df[col] < 0).sum()
    print(f"Coluna: {col} | Negativos: {qtd}")
```

```python
df_numerico = df.select_dtypes(include=[np.number])

colunas_com_inf = df_numerico.columns[np.isinf(df_numerico).any()].tolist()

colunas_com_nan = df_numerico.columns[df_numerico.isna().any()].tolist()

print(f"Colunas com Infinito: {colunas_com_inf}")
print(f"Colunas com NaN: {colunas_com_nan}")
```

```python
correlacao = df['Fwd Header Length'].corr(df['Fwd Header Length.1'])
print(correlacao)
```

```python
unicos_total = df.nunique()

estaticas_total = unicos_total[unicos_total == 1]
if estaticas_total.empty:
    print("Não há colunas estáticas no dataset inteiro.")
else:
    print(f"Features estáticas no TOTAL ({len(estaticas_total)} colunas):")
    for col in estaticas_total.index:
        valor = df[col].iloc[0]
        print(f"{col}: {valor}")
```
