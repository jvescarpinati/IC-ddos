```python
df_numerico = df.select_dtypes(include=[np.number])

colunas_com_inf = df_numerico.columns[np.isinf(df_numerico).any()].tolist()

colunas_com_nan = df_numerico.columns[df_numerico.isna().any()].tolist()

print(f"Colunas com Infinito: {colunas_com_inf}")
print(f"Colunas com NaN: {colunas_com_nan}")
```

```python
unicos_total = df.nunique()

estaticas_total = unicos_total[unicos_total == 1]
if estaticas_total.empty:
    print("Não há colunas estáticas no dataset inteiro.")
else:
    print(f"Features estáticas no TOTAL ({len(estaticas_total)} colunas):")
    for col in estaticas_total.index:
        valor = df[col].iloc[0]
        print(f"{col}: {valor}")
```
