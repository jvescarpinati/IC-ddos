```python
import pandas as pd
import numpy as np
```

```python
df_ssdp = pd.read_csv('../../../datasets/CICDDOS2019/classe_unica/fluxo_ssdp.csv')
df_ssdp.columns = df_ssdp.columns.str.strip()
```

```python
print(df_ssdp['Label'].value_counts())
```

```python
colunas_remover = [
    'Unnamed: 0', 
    'Flow ID', 
    'Source IP', 
    'Source Port', 
    'Destination IP', 
    'Destination Port', 
    'Protocol', 
    'Timestamp', 
    'Flow Duration',
    'SimillarHTTP',
    'Bwd PSH Flags',
    'Bwd URG Flags',
    'Fwd URG Flags',
    'FIN Flag Count',
    'PSH Flag Count',
    'ECE Flag Count',
    'Fwd Avg Bytes/Bulk',
    'Fwd Avg Packets/Bulk',
    'Fwd Avg Bulk Rate',
    'Bwd Avg Bytes/Bulk',
    'Bwd Avg Packets/Bulk',
    'Bwd Avg Bulk Rate',
    'Fwd Header Length.1',
    'Init_Win_bytes_backward',
    'Init_Win_bytes_forward'
]

df_ssdp.drop(columns=colunas_remover, inplace=True)

df_ssdp['Label'] = df_ssdp['Label'].replace('DrDoS_SSDP', 1)
df_ssdp['Label'] = df_ssdp['Label'].astype(int)
```

```python
df_ssdp.replace([np.inf, -np.inf], np.nan, inplace=True)

df_ssdp.dropna(subset=['Flow Bytes/s', 'Flow Packets/s'], inplace=True)

df_ssdp.reset_index(drop=True, inplace=True)

colunas_numericas = df_ssdp.select_dtypes(include=np.number).columns

df_ssdp = df_ssdp[(df_ssdp[colunas_numericas] >= 0).all(axis=1)]

df_ssdp.reset_index(drop=True, inplace=True)
```

```python
df_ssdp.to_csv('../../../datasets/tratados/dataset_ssdp_tratado.csv', index=False)

print("Dataset salvo")
```
