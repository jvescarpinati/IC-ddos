```python
import pandas as pd
import numpy as np
```

```python
df_udp = pd.read_csv('../../../datasets/CICDDOS2019/classe_unica/fluxo_drdos_udp.csv')
df_udp.columns = df_udp.columns.str.strip()
df_udp['Label'].value_counts()
```

```python
colunas_remover = [
    'Unnamed: 0', 
    'Flow ID', 
    'Source IP', 
    'Source Port', 
    'Destination IP', 
    'Destination Port', 
    'Protocol', 
    'Timestamp', 
    'Flow Duration',
    'SimillarHTTP',
    'Bwd PSH Flags',
    'Bwd URG Flags',
    'Fwd URG Flags',
    'FIN Flag Count',
    'PSH Flag Count',
    'ECE Flag Count',
    'Fwd Avg Bytes/Bulk',
    'Fwd Avg Packets/Bulk',
    'Fwd Avg Bulk Rate',
    'Bwd Avg Bytes/Bulk',
    'Bwd Avg Packets/Bulk',
    'Bwd Avg Bulk Rate',
    'Fwd Header Length.1',
    'Init_Win_bytes_backward',
    'Init_Win_bytes_forward'
]

df_udp.drop(columns=colunas_remover, inplace=True)

df_udp['Label'] = df_udp['Label'].replace('DrDoS_UDP', 1)
df_udp['Label'] = df_udp['Label'].astype(int)

df_udp.replace([np.inf, -np.inf], np.nan, inplace=True)
df_udp.dropna(subset=['Flow Bytes/s', 'Flow Packets/s'], inplace=True)
df_udp.reset_index(drop=True, inplace=True)

colunas_numericas = df_udp.select_dtypes(include=np.number).columns
df_udp = df_udp[(df_udp[colunas_numericas] >= 0).all(axis=1)]
df_udp.reset_index(drop=True, inplace=True)
```

```python
unicos_total = df_udp.nunique()
estaticas_total = unicos_total[unicos_total == 1]
if estaticas_total.empty:
    print("Não há colunas estáticas no dataset inteiro.")
else:
    print(f"Features estáticas no TOTAL ({len(estaticas_total)} colunas):")
    for col in estaticas_total.index:
        valor = df_udp[col].iloc[0]
        print(f"{col}: {valor}")
```

```python
df_udp.to_csv('../../../datasets/tratados/dataset_udp_tratado.csv', index=False)

print("Dataset salvo")
```
