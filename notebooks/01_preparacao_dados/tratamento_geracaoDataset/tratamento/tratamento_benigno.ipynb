```python
import pandas as pd
import numpy as np
```

```python
df_limpo = pd.read_csv('../../../datasets/CICDDOS2019/classe_unica/dataset_benigno.csv')
pd.set_option('display.max_columns', None)
df_limpo.head()
```

```python
df_limpo['Label'].value_counts()
```

```python
df_limpo.info()
```

```python
df_numerico = df_limpo.select_dtypes(include=[np.number])

colunas_com_inf = df_numerico.columns[np.isinf(df_numerico).any()].tolist()

colunas_com_nan = df_numerico.columns[df_numerico.isna().any()].tolist()

print(f"Colunas com Infinito: {colunas_com_inf}")
print(f"Colunas com NaN: {colunas_com_nan}")
```

```python
unicos_total = df_limpo.nunique()

estaticas_total = unicos_total[unicos_total == 1]
if estaticas_total.empty:
    print("Não há colunas estáticas no dataset inteiro.")
else:
    print(f"Features estáticas no TOTAL ({len(estaticas_total)} colunas):")
    for col in estaticas_total.index:
        valor = df_limpo[col].iloc[0]
        print(f"{col}: {valor}")
```

```python
colunas_remover = [
    'Unnamed: 0', 
    'Flow ID', 
    'Source IP', 
    'Source Port', 
    'Destination IP', 
    'Destination Port', 
    'Protocol', 
    'Timestamp', 
    'Flow Duration',
    'SimillarHTTP',
    'Bwd PSH Flags',
    'Bwd URG Flags',
    'Fwd URG Flags',
    'FIN Flag Count',
    'PSH Flag Count',
    'ECE Flag Count',
    'Fwd Avg Bytes/Bulk',
    'Fwd Avg Packets/Bulk',
    'Fwd Avg Bulk Rate',
    'Bwd Avg Bytes/Bulk',
    'Bwd Avg Packets/Bulk',
    'Bwd Avg Bulk Rate',
    'Fwd Header Length.1',
    'Init_Win_bytes_backward',
    'Init_Win_bytes_forward'
]

df_limpo.drop(columns=colunas_remover, inplace=True)
```

```python
df_limpo['Label'] = df_limpo['Label'].replace('BENIGN', 0)
df_limpo['Label'] = df_limpo['Label'].astype(int)
```

```python
df_limpo.info()
```

```python
df_limpo.replace([np.inf, -np.inf], np.nan, inplace=True)

df_limpo.dropna(subset=['Flow Bytes/s', 'Flow Packets/s'], inplace=True)

df_limpo.reset_index(drop=True, inplace=True)
```

```python

colunas_numericas = df_limpo.select_dtypes(include=np.number).columns

df_limpo = df_limpo[(df_limpo[colunas_numericas] >= 0).all(axis=1)]

df_limpo.reset_index(drop=True, inplace=True)
```

```python
df_limpo['Label'].value_counts()
```

```python
df_limpo.to_csv('../../../datasets/tratados/dataset_benigno_limpo.csv', index=False)

print("Dataset salvo")
```
