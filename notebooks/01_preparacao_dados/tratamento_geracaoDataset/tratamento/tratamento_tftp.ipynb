```python
import pandas as pd
import numpy as np
```

```python
df_tftp = pd.read_csv('../../../datasets/CICDDOS2019/classe_unica/fluxo_tftp.csv')
df_tftp.columns = df_tftp.columns.str.strip()

print(df_tftp['Label'].value_counts())
```

```python
colunas_remover = [
    'Unnamed: 0', 
    'Flow ID', 
    'Source IP', 
    'Source Port', 
    'Destination IP', 
    'Destination Port', 
    'Protocol', 
    'Timestamp', 
    'Flow Duration',
    'SimillarHTTP',
    'Bwd PSH Flags',
    'Bwd URG Flags',
    'Fwd URG Flags',
    'FIN Flag Count',
    'PSH Flag Count',
    'ECE Flag Count',
    'Fwd Avg Bytes/Bulk',
    'Fwd Avg Packets/Bulk',
    'Fwd Avg Bulk Rate',
    'Bwd Avg Bytes/Bulk',
    'Bwd Avg Packets/Bulk',
    'Bwd Avg Bulk Rate',
    'Fwd Header Length.1',
    'Init_Win_bytes_backward',
    'Init_Win_bytes_forward'
]

df_tftp.drop(columns=colunas_remover, inplace=True)

df_tftp['Label'] = df_tftp['Label'].replace('TFTP', 1)
df_tftp['Label'] = df_tftp['Label'].astype(int)

```

```python
df_tftp.info()
```

```python
df_tftp.replace([np.inf, -np.inf], np.nan, inplace=True)

df_tftp.dropna(subset=['Flow Bytes/s', 'Flow Packets/s'], inplace=True)

df_tftp.reset_index(drop=True, inplace=True)
```

```python
colunas_numericas = df_tftp.select_dtypes(include=np.number).columns

df_tftp = df_tftp[(df_tftp[colunas_numericas] >= 0).all(axis=1)]

df_tftp.reset_index(drop=True, inplace=True)
```

```python
df_tftp.to_csv('../../../datasets/tratados/dataset_tftp_tratado.csv', index=False)

print("Dataset salvo")
```
