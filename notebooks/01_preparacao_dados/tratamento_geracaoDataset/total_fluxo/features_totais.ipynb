```python
import pandas as pd
import numpy as np
```

```python
df_benigno = pd.read_csv('../../../datasets/tratados/dataset_benigno_limpo.csv')

df_udp = pd.read_csv('../../../datasets/tratados/dataset_udp_tratado.csv')

df_syn = pd.read_csv('../../../datasets/tratados/dataset_syn_tratado.csv')

df_ntp = pd.read_csv('../../../datasets/tratados/dataset_ntp_tratado.csv')

df_ldap = pd.read_csv('../../../datasets/tratados/dataset_ldap_tratado.csv')

df_dns = pd.read_csv('../../../datasets/tratados/dataset_dns_tratado.csv')

df_netbios = pd.read_csv('../../../datasets/tratados/dataset_netbios_tratado.csv')

df_ssdp = pd.read_csv('../../../datasets/tratados/dataset_ssdp_tratado.csv')

df_mssql = pd.read_csv('../../../datasets/tratados/dataset_mssql_tratado.csv')

df_tftp = pd.read_csv('../../../datasets/tratados/dataset_tftp_tratado.csv')

df_snmp = pd.read_csv('../../../datasets/tratados/dataset_snmp_tratado.csv')
```

```python
n_amostras = 5240
seed = 42

df_final = pd.concat([
    df_benigno,
    df_udp.sample(n=n_amostras, random_state=seed),
    df_syn.sample(n=n_amostras, random_state=seed),
    df_ntp.sample(n=n_amostras, random_state=seed),
    df_ldap.sample(n=n_amostras, random_state=seed),
    df_dns.sample(n=n_amostras, random_state=seed),
    df_netbios.sample(n=n_amostras, random_state=seed),
    df_ssdp.sample(n=n_amostras, random_state=seed),
    df_mssql.sample(n=n_amostras, random_state=seed),
    df_tftp.sample(n=n_amostras, random_state=seed),
    df_snmp.sample(n=n_amostras, random_state=seed)
])
df_final = df_final.sample(frac=1, random_state=seed).reset_index(drop=True)
```

```python
df_final['Label'].value_counts()
```

```python
df_final.replace([np.inf, -np.inf], np.nan, inplace=True)

df_final.dropna(subset=['Flow Bytes/s', 'Flow Packets/s'], inplace=True)

df_final.reset_index(drop=True, inplace=True)
```

```python
colunas_numericas = df_final.select_dtypes(include=np.number).columns

df_final = df_final[(df_final[colunas_numericas] >= 0).all(axis=1)]

df_final.reset_index(drop=True, inplace=True)
```

```python
unicos_total = df_final.nunique()

estaticas_total = unicos_total[unicos_total == 1]
if estaticas_total.empty:
    print("Não há colunas estáticas no dataset inteiro.")
else:
    print(f"Features estáticas no TOTAL ({len(estaticas_total)} colunas):")
    for col in estaticas_total.index:
        valor = df_final[col].iloc[0]
        print(f"{col}: {valor}")
```
